```python
from transformers import AutoProcessor, SeamlessM4Tv2Model

processor = AutoProcessor.from_pretrained("facebook/seamless-m4t-v2-large")
model = SeamlessM4Tv2Model.from_pretrained("facebook/seamless-m4t-v2-large")

#from transformers import AutoProcessor, SeamlessM4Tv2ForTextToText
#model = SeamlessM4Tv2ForTextToText.from_pretrained("facebook/seamless-m4t-v2-large")
#processor = AutoProcessor.from_pretrained("facebook/seamless-m4t-v2-large")
```

```python
# Define the credentials for the SeamlessM4Tv2 model
src_lang = "por"
tgt_lang = "eng"
```

```python
#Replace 'path/to/file.csv' with the actual path to your CSV file
path_csv_file = '/content/drive/MyDrive/Dados/Traducao/input_active_ing_translate_meta.csv'
```

```python
df = pd.read_csv(path_csv_file)
```

```python
df.head()
```

```python
# Function to translate text

def translate_text(text_ori):
    text_inputs = processor(text=text_ori, src_lang=src_lang, return_tensors="pt")
    output_tokens = model.generate(**text_inputs, tgt_lang=tgt_lang, generate_speech=False)
    translated_text = processor.decode(output_tokens[0].tolist()[0], skip_special_tokens=True)

    return translated_text
```

```python
# Loop to translate line by line and print
for index, row in df.iterrows():
    texto_original = row["active_ingredient"]

    texto_traduzido = translate_text(texto_original)
    print(f"Original: {texto_original} | Traduzido: {texto_traduzido}")

    # Atualizar o DataFrame com as traduções
    df.at[index, "active_ingredient_eng"] = texto_traduzido
```
